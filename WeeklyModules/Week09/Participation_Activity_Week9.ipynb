```python
import os
import json
import requests
```

```python
searchTerm = 'flowers'

folderName = 'downloaded_images'
os.makedirs(folderName, exist_ok=True)
```

```python
print('Searching Flickr for ' + searchTerm + '...')

url = 'https://www.flickr.com/services/feeds/photos_public.gne'

params = {
    'tags': searchTerm,
    'format': 'json',
    'nojsoncallback': 1
}

res = requests.get(url, params=params, timeout=30)
res.raise_for_status()
```

```python
feedText = res.text.replace("\\'", "'")
data = json.loads(feedText)
```

```python
count = 0

for item in data['items']:
    imageUrl = item['media']['m']
    filename = os.path.join(folderName, os.path.basename(imageUrl))

    try:
        print('Downloading ' + imageUrl)

        imageRes = requests.get(imageUrl, timeout=30)
        imageRes.raise_for_status()

        # Save the image in binary mode.
        with open(filename, 'wb') as imageFile:
            for chunk in imageRes.iter_content(100000):
                imageFile.write(chunk)

        count = count + 1

    except requests.RequestException:
        print('Could not download ' + imageUrl)
```

```python
print('Done. Downloaded ' + str(count) + ' images.')
print('Images saved in: ' + os.path.abspath(folderName))
```
