```python
df = pd.read_csv('/content/MSSP6070/data/InternetSales.csv', encoding = 'latin1')
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
df["Year"] = df["OrderDate"].dt.year
yearly_sales = (df.groupby("Year")["SalesAmount"].sum())
yearly_sales = yearly_sales.sort_index()
```

```python
yearly_sales.plot(kind="bar")
plt.title("Yearly Sales")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.ticklabel_format(style="plain", axis="y")
plt.show
plt.savefig("/content/MSSP6070/data/yearly_sales.pdf")
```
