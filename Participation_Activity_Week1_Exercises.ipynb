## Exercise 6: Calculating more statistics
### range

```python
range = max(salaries) - min(salaries)
print(range)
```

### coefficient of variation

```python
coeff_of_variation = (sample_std_deviation / mean_salary)
print(coeff_of_variation)
```

### interquartile range

```python
salaries.sort()
q1_left = salaries[24]
q1_right = salaries[25]
q1 = (q1_left + q1_right) / 2
q3_left = salaries[74]
q3_right = salaries[75]
q3 = (q3_left + q3_right) / 2
IQR = q3 - q1
print(IQR)
```

### quartile coefficent of dispersion

```python
QCD = (q3 - q1) / (q3 + q1)
print(QCD)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_max_scaled = []
for salary in salaries:
  scaled = (salary - min(salaries)) / range
  min_max_scaled.append(scaled)
print(min_max_scaled)
```

### standardizing

```python
standardized_salaries = []
for salary in salaries:
  z_score = (salary - mean_salary) / sample_std_deviation
  standardized_salaries.append(z_score)
print(standardized_salaries)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
from statistics import mean, stdev
mean_salary, std_salary = mean(salaries), stdev(salaries)

standardized = [(x - mean_salary) / std_salary for x in salaries]
standardized[:5]

import numpy as np
np.cov(min_max_scaled, standardized)
```

### Pearson correlation coefficient ($\rho$)

```python
from statistics import correlation
x = min_max_scaled
y = standardized
pearson = correlation(x, y)
print(pearson)
```
