2. f)

```python
faang = pd.read_csv('/content/MSSP6070/data/faang.csv')
```

```python
faang['date'] = pd.to_datetime(faang['date'])
faang['volume'] = faang['volume'].astype('int')
faang.dtypes
```

```python
faang = faang.sort_values(
    by=['date', 'ticker']
)
faang.head()
```

2. g)

```python
top_7_volume = faang.nlargest(7, 'volume')
top_7_volume
```

2. h)

```python
melted = pd.melt(faang, ['date', 'ticker'])
faang_melted = melted.sort_values(
    by=['date', 'ticker']
)
faang_melted.head()
```
