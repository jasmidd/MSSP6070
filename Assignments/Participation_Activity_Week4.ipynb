```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/MSSP6070/Modules/Week04/parsed.csv')
```

2a.

```python
japan_mb = df[
    (df["parsed_place"] == "Japan") &
    (df["magType"] == "mb")
]
percentile_95 = japan_mb["mag"].quantile(0.95)
print(f"95th percentile of earthquake magnitude in Japan using magType of 'mb': \n{percentile_95}")
```

2b.

```python
indonesia_eq = df[
    (df["parsed_place"] == "Indonesia") &
    (df["type"] == "earthquake")
]
percentage_tsunami = ((indonesia_eq["tsunami"] == 1).sum()) / len(indonesia_eq) * 100
print(f"Percentage of earthquakes in Indonesia coupled with tsunanmis: \n{percentage_tsunami}")
```
