```python
def collatz(number):
  if number % 2 == 0:
    value = number // 2
    print(value, end=' ')
    return value
  elif number % 2 == 1:
    value = 3 * number + 1
    print(value, end=' ')
    return value

print('Enter a number: ')
number = int(input())
while number != 1:
  number = collatz(number)
```
